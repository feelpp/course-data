# Data lifecycle and first audit

This lab turns a small, deliberately flawed table into an executable data-quality report.

```python
import pandas as pd

observations = pd.DataFrame(
    {
        "sample_id": [101, 102, 102, 104],
        "temperature_c": [21.4, None, 19.8, -280.0],
        "station": ["north", "north", "north", "south"],
    }
)
observations
```

## Task

Complete `audit_frame`. Report the row count, duplicated identifiers, missing temperatures, and temperatures below absolute zero.

```python
def audit_frame(frame: pd.DataFrame) -> dict[str, int]:
    """Return reproducible counts for the required data-quality checks."""
    return {
        "rows": len(frame),
        "duplicate_ids": int(frame["sample_id"].duplicated(keep=False).sum()),
        "missing_temperature": int(frame["temperature_c"].isna().sum()),
        "below_absolute_zero": int((frame["temperature_c"] < -273.15).sum()),
    }
```

```python
audit = audit_frame(observations)
audit
```

```python
assert audit == {
    "rows": 4,
    "duplicate_ids": 2,
    "missing_temperature": 1,
    "below_absolute_zero": 1,
}
```

**Instructor note.** Ask students why two rows are counted for one duplicated identifier and why the impossible temperature should block analysis rather than be silently clipped.
